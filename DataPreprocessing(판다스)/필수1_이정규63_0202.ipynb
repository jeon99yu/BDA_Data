- 긍정과 부정 두 개 비교해서 상위 5개 리스트 공유

```python
import re
import nltk
import pandas as pd
import numpy as np
from nltk.tokenize import sent_tokenize,word_tokenize,WordPunctTokenizer,RegexpTokenizer
from sklearn.feature_extraction.text import CountVectorizer
from nltk.corpus import stopwords
from collections import Counter
```

```python
df = pd.read_csv(r'C:\Users\user\BDA_Data\Dataset\movie_rv.csv')
df
```

```python
df['document'].str.split(' ', n=3, expand=True)
```

```python
# 긍정과 부정 리뷰 데이터 정의
pos_reviews = df[df["label"] == 1]["document"].dropna().tolist()
neg_reviews = df[df["label"] == 0]["document"].dropna().tolist()

# CountVectorizer를 활용한 단어 빈도 분석
vectorizer = CountVectorizer(stop_words=None, max_features=1000)  # 상위 1000개 단어만 고려

# 긍정과 부정 리뷰를 벡터화
pos_matrix = vectorizer.fit_transform(pos_reviews)
neg_matrix = vectorizer.fit_transform(neg_reviews)

# 단어 목록 가져오기 (구버전 대응)
try:
    words = vectorizer.get_feature_names_out()  # 최신 버전
except AttributeError:
    words = vectorizer.get_feature_names()  # 구버전 대응

# 단어 빈도 계산
pos_word_counts = pos_matrix.toarray().sum(axis=0)
neg_word_counts = neg_matrix.toarray().sum(axis=0)

# 상위 5개 키워드 추출
top_pos_words = sorted(zip(words, pos_word_counts), key=lambda x: x[1], reverse=True)[:5]
top_neg_words = sorted(zip(words, neg_word_counts), key=lambda x: x[1], reverse=True)[:5]

top_pos_words, top_neg_words
```
