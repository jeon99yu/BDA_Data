```python
## 지도에 좌표를 찍어서 확인해 보자!

def cal_euc_dis(row1, row2):
    pnt1 = np.array([row1['latitude'],row1['longitude']])
    pnt2 = np.array([row2['latitude'],row2['longitude']])
    return np.linalg.norm(pnt1- pnt2)
```

```python
## 지도에 기준점이 필요하다. 
## 기준점은 평균으로 잡고 각 거리를 계산해 보자

map_center =[df_ydp['latitude'].mean(), df_ydp['longitude'].mean()]

#folium 시각화를 해서 지도 위치 시각화

m =folium.Map(location=map_center, zoom_start=10)
```
